```python
import os
import sys
import time
from pathlib import Path

import pandas as pd
import polars as pl

THIS_PATH = Path(os.getcwd())
ROOT_PATH = THIS_PATH.parent.absolute()
DATA_PATH = ROOT_PATH / 'data'
RAW_PATH = DATA_PATH / 'raw'

accepted_loans_filename = 'accepted_2007_to_2018Q4.csv'
rejected_loans_filename = 'rejected_2007_to_2018Q4.csv'

accepted_loans_filepath = RAW_PATH / accepted_loans_filename / accepted_loans_filename
rejected_loans_filepath = RAW_PATH / rejected_loans_filename / rejected_loans_filename

if not accepted_loans_filepath.exists():
    raise FileNotFoundError

if not rejected_loans_filepath.exists():
    raise FileNotFoundError
```

```python
tik_df_accepted_pd = time.perf_counter()
df_accepted_pd = pd.read_csv(accepted_loans_filepath, dtype=str)
tok_df_accepted_pd = time.perf_counter()

time_df_accepted_pd = tok_df_accepted_pd - tik_df_accepted_pd
print(time_df_accepted_pd)
```

```python
tik_df_accepted_pl = time.perf_counter()
df_accepted_pl = pl.read_csv(accepted_loans_filepath, infer_schema=False)
tok_df_accepted_pl = time.perf_counter()

time_df_accepted_pl = tok_df_accepted_pl - tik_df_accepted_pl
print(time_df_accepted_pl)
```
