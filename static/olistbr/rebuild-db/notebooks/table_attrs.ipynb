```python
import os
from pathlib import Path

import pandas as pd
from pandas import DataFrame
from icecream import ic

from src.utils.infer import infer_dtypes
from src.utils.paths import RAW_PATH

raw_files = os.listdir(RAW_PATH)

filepaths: list[Path] = []
dfs: dict[str, DataFrame] = dict()

for raw_file in raw_files:
    filepath = RAW_PATH / raw_file
    filepaths.append(filepath)

    if not filepath.exists():
        raise FileNotFoundError

    dfs[filepath.name] = pd.read_csv(filepath, dtype=object)
```

```python
pd.set_option("display.max_columns", None)

filename = "olist_" + "closed_deals" + "_dataset.csv"
df = dfs[filename]

df_attributes = infer_dtypes(df)
df_attributes
```
